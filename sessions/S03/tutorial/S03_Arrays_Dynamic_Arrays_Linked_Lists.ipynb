## Linked-list setup

Run this cell. A valid singly linked list starts at head, visits every live node exactly once and ends at None.

````python
from dataclasses import dataclass

@dataclass(eq=False)
class Node:
    value: int
    next: "Node | None" = None

def values_from(head):
    values, seen = [], set()
    while head is not None:
        assert head not in seen, "Cycle detected"
        seen.add(head)
        values.append(head.value)
        head = head.next
    return values
````
