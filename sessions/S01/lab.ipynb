```python
def linear_search(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return -1

def has_duplicate(values):
    for first in range(len(values)):
        for second in range(first + 1, len(values)):
            if values[first] == values[second]:
                return True
    return False

assert linear_search([], 4) == -1
assert linear_search([4], 4) == 0
assert linear_search([9, 4, 4, 1], 4) == 1
assert linear_search([9, 4, 4, 1], 7) == -1
assert not has_duplicate([])
assert not has_duplicate([4])
assert has_duplicate([4, 4])
assert not has_duplicate([3, 8, 5])
print('Notebook edge-case checks passed')
```

## Predict, then observe
For an absent target, predict the equality-check count at each `n`. For distinct input, predict the number of pairwise comparisons. Time the complete function call on deterministic inputs. These observations are not a complexity proof and raw Python/C++ times are not directly comparable.

```python
import platform
import statistics
import time

def median_ns(call, repeats=11):
    call()  # warm-up
    observations = []
    for _ in range(repeats):
        start = time.perf_counter_ns()
        result = call()
        observations.append(time.perf_counter_ns() - start)
        assert result is not None
    return int(statistics.median(observations))

print('Python:', platform.python_version(), 'Platform:', platform.platform())
print('Timer: perf_counter_ns; one warm-up; median of 11; deterministic inputs')
print('n, predicted absent checks, search median ns, predicted distinct-pair checks, pair median ns')
for n in (16, 64, 256, 512):
    values = list(range(n))
    search_time = median_ns(lambda: linear_search(values, -1))
    pair_time = median_ns(lambda: has_duplicate(values))
    print(n, n, search_time, n * (n - 1) // 2, pair_time, sep=', ')
```
